```python
import time

def knapsack(weights, profits, capacity):
    n = len(weights)

    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(capacity + 1):
            if weights[i - 1] <= w:
                dp[i][w] = max(
                    profits[i - 1] + dp[i - 1][w - weights[i - 1]],
                    dp[i - 1][w]
                )
            else:
                dp[i][w] = dp[i - 1][w]

    return dp[n][capacity]


n = int(input("Enter number of items: "))

weights = []
profits = []

print("Enter weights:")
for i in range(n):
    weights.append(int(input()))

print("Enter profits:")
for i in range(n):
    profits.append(int(input()))

capacity = int(input("Enter knapsack capacity: "))

start = time.perf_counter()

max_profit = knapsack(weights, profits, capacity)

end = time.perf_counter()

execution_time = end - start

print("\nMaximum Profit:", max_profit)
print("Execution Time:", execution_time, "seconds")
print("Time Complexity: O(n * W)")
print("Space Complexity: O(n * W)")
```

Summary

Dynamic Programming is used to solve the 0/1 Knapsack problem by dividing it into smaller subproblems and storing their solutions in a table. For each item and capacity, the algorithm decides whether to include or exclude the item. This avoids repeated calculations and efficiently finds the maximum possible profit.

Conclusion

The 0/1 Knapsack problem was successfully implemented using Dynamic Programming in Python. The program accepts the number of items, weights, profits, and knapsack capacity as user input. It calculates the maximum profit and measures the execution time. The time complexity is O(n × W) and the space complexity is O(n × W).
