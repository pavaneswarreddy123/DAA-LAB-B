```python
from collections import deque
import time

# ---------------- BFS ----------------
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        vertex = queue.popleft()
        print(vertex, end=" ")

        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


# ---------------- DFS ----------------
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)
    print(start, end=" ")

    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)


# ---------------- User Input ----------------

graph = {}

n = int(input("Enter number of vertices: "))

for i in range(n):
    vertex = input(f"Enter vertex {i + 1}: ")
    graph[vertex] = []

e = int(input("Enter number of edges: "))

for i in range(e):
    u, v = input(f"Enter edge {i + 1} (u v): ").split()

    graph[u].append(v)
    graph[v].append(u)       # Remove this line for directed graph


start = input("Enter starting vertex: ")

# ---------------- BFS Execution Time ----------------

start_time = time.perf_counter()

print("\nBFS Traversal:")
bfs(graph, start)

bfs_time = time.perf_counter() - start_time

# ---------------- DFS Execution Time ----------------

start_time = time.perf_counter()

print("\nDFS Traversal:")
dfs(graph, start)

dfs_time = time.perf_counter() - start_time

# ---------------- Results ----------------

print("\n\nExecution Time:")
print("BFS:", bfs_time, "seconds")
print("DFS:", dfs_time, "seconds")

print("\nComplexity:")
print("BFS Time Complexity  : O(V + E)")
print("BFS Space Complexity : O(V)")

print("DFS Time Complexity  : O(V + E)")
print("DFS Space Complexity : O(V)")
```
