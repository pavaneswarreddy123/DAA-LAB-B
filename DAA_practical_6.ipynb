```python
import time

def matrix_chain_order(p, n):
    dp = [[0] * n for _ in range(n)]

    for length in range(2, n):
        for i in range(1, n - length + 1):
            j = i + length - 1
            dp[i][j] = float('inf')

            for k in range(i, j):
                cost = (dp[i][k] +
                        dp[k + 1][j] +
                        p[i - 1] * p[k] * p[j])

                if cost < dp[i][j]:
                    dp[i][j] = cost

    return dp[1][n - 1]


print("Matrix Chain Multiplication")

n = int(input("Enter number of matrices: "))

p = []

print("Enter matrix dimensions:")
for i in range(n):
    if i == 0:
        r = int(input("Enter rows of matrix 1: "))
        c = int(input("Enter columns of matrix 1: "))
        p.append(r)
        p.append(c)
    else:
        r = p[-1]
        c = int(input(f"Enter columns of matrix {i + 1}: "))
        p.append(c)

start = time.perf_counter()

result = matrix_chain_order(p, n + 1)

end = time.perf_counter()

print("\nMinimum number of scalar multiplications:", result)
print("Execution time:", end - start, "seconds")
print("Time Complexity: O(n^3)")
print("Space Complexity: O(n^2)")
```

Summary

Matrix Chain Multiplication is an optimization problem solved using Dynamic Programming. It finds the best order for multiplying a sequence of matrices so that the number of scalar multiplications is minimized. The algorithm stores solutions to smaller subproblems and uses them to solve larger problems efficiently.

Conclusion

The Matrix Chain Multiplication algorithm was successfully implemented using Dynamic Programming. The minimum number of scalar multiplications was calculated for the given matrices. The algorithm has O(n³) time complexity and O(n²) space complexity, making it more efficient than checking all possible multiplication orders.
